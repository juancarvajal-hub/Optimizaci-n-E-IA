**Parte 1 : OPTIMIZACIÓN DE FUNCIONES USANDO BÚSQUEDA DE GRILLA Y BÚSQUEDA ALEATORIA**

```python
#--------------------- Importación de librerias -----------------------#
########################################################################

import numpy as np 
import pandas as pd
import matplotlib.pyplot as plt
from numpy.random import rand
```

**Expresión de la función :** Esta función es la que nos va a interesas optimizar.

$f(x, y) = 0.26(x^2 + y^2 - 0.48xy)$ para $x, y \in [-10, 10]$.

```python
# Definimos la función objetivo
def test_function1(x,y):    
    return 0.26*(x**2+y**2-0.48*x*y)
```

Para visulizar la función primero debemos encontrar muestras de ella en el rango especifico de sus variables.

```python
xi = np.arange(-10,10,0.1) # Generamos valores para la variable x
yi = np.arange(-10,10,0.1) # Generamos valores aleatorios para la variable y
x,y = np.meshgrid(xi,yi) # Luego encontramos todo el  conjunto de combinaciones de las varaibles x , y
feval =  test_function1(x,y)
```

**Visualización de la función objetivo**

```python
plt.contourf(x, y, feval, levels=50, cmap='jet')
plt.colorbar()
plt.show()
```

### **Búsqueda Por Grilla**

```python
# Generemos la grilla de busqueda

x_sample = np.arange(-7,7,0.5)
y_sample = np.arange(-7,7,0.5)

xt, yt =np.meshgrid(x_sample,y_sample)

#Calculamos la funcion
feval_t = test_function1(xt,yt)
```

```python
#Busqueda en grilla
f_opt = feval_t[0][0]
best_x = 0
best_y = 0
###
for i in range(feval_t.shape[0]):
    for j in range(feval_t.shape[1]):
        if feval_t[i][j] < f_opt:
            f_opt = feval_t[i][j]
            best_x = i
            best_y = j


print('Punto optimo: x:',xt[best_x][best_y],'y:',yt[best_x][best_y])
print('Estos puntos dan como valor optimo:' ,f_opt)
```

**Grafica de la función con los puntos de la grilla y el punto Optimo que arrojó el algoritmo**

```python
plt.contourf(x, y, feval, levels=50, cmap='jet')
plt.scatter(xt,yt,c='black',marker='.')
plt.scatter(xt[best_x][best_y],yt[best_x][best_y],c='red',s=10)
plt.colorbar()
plt.show()
```

### **Búsqueda Aleatoria**

```python
#Generamos los puntos aleatorio
###
r_min , r_max = -7,7
x_sample = r_min + rand(500)*(r_max-r_min)
y_sample = r_min + rand(500)*(r_max-r_min)

x_t , y_t = np.meshgrid(x_sample,y_sample)
#Calculamos la funcion
feval_t=test_function1(x_t,y_t)
```

```python
#Busqueda valor optimo
f_opt = feval_t[0][0]
for i in range(feval_t.shape[0]):
  for j in range(feval_t.shape[1]):
    if feval_t[i][j]<f_opt:
      f_opt=feval_t[i][j]
      best_x = i
      best_y = j

print('Punto optimo: x:',x_t[best_x][best_y],'y:',y_t[best_x][best_y])
print('Valor optimo:' ,f_opt)
```

```python
#Grafica
plt.contourf(x, y, feval, levels=50, cmap='jet')
plt.scatter(x_sample,y_sample,c='black',marker='.')
plt.scatter(x_t[best_x][best_y],y_t[best_x][best_y],c='red',marker='.')
plt.colorbar()
plt.show()
```

**Expresión de la función :** Esta función es la que nos va a interesas optimizar.

$f(x, y) = -\cos(x)\cos(y)\exp\{-((x-\pi)^2 + (y-\pi)^2)\} \quad \text{para } x, y \in [-10, 10]$.

```python
# Definimos la función objetivo
def test_function2(x,y):    
    return -np.cos(x) * np.cos(y) * np.exp(-((x - np.pi)**2 + (y - np.pi)**2))
```

```python
xi = np.arange(-10,10,0.1) # Generamos valores para la variable x
yi = np.arange(-10,10,0.1) # Generamos valores aleatorios para la variable y
x,y = np.meshgrid(xi,yi) # Luego encontramos todo el  conjunto de combinaciones de las varaibles x , y
feval =  test_function2(x,y)
```

**Visualización de la función objetivo**

```python
plt.contourf(x, y, feval, levels=50, cmap='jet')
plt.colorbar()
plt.xlim(1.5, 4.5)  
plt.ylim(1.5, 4.5) 
plt.show()
```
